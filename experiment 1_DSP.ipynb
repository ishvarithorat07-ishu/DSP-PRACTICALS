```python
#Arithmetic Operators
a = 15
b = 4
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
#Relational or Comaprison operators
a = 13
b = 33
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
#Logical Operators
a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

```python
#Bitwise Operators
a = 10
b = 4
print(a & b)
print(a | b)
print(~a)
print(a ^ b)
print(a >> 2)
print(a << 2)
```

```python
#Assignment Operators
a = 10
b = a
print(b)
b += a
print(b)
b -= a
print(b)
b *= a
print(b)
b <<= a
print(b)
```

```python
#Identity Operators
a = 10
b = 20
c = a
print(a is not b)
print(a is c)
```

```python
#Membership Operators
x = 24
y = 20
my_list = [10, 20, 30, 40, 50]
if (x not in my_list):
print("x is NOT present in given list")
else:
print("x is present in given list")
if (y in my_list):
print("y is present in given list")
else:
print("y is NOT present in given list")
```

```python
#Numeric Data Type
a = 5
b = 5.0
c = 2 + 4j
print(type(a))
print(type(b))
print(type(c))
```

```python
#Sequence Data Type---> String
s = 'Welcome to the Geeks World'
print(s)
print(type(s))
# access string with index
print(s[1])
print(s[-1])
```

```python
#Sequence Data Type---> List
a = [1, 2, 3]
print(a)
b = ["Geeks", "For", "Geeks", 4, 5]
print(b[3])
print(b[-3])
```

```python
#Sequence Data Type---> Tuple
t1 = (1,)
print(type(t1))
t2 = ('Geeks', 'For', 'Geeks', 1, 2)
print(t2[3])
print(t2[-3])
```

```python
#Boolean Data Type
print(type(True))
print(type(False))
```

```python
#Set Data Type
s1 = {"a", "a", "b", "c", "b"}
print(s1)
s2 = {"Geeks", "For", "Geeks"}
for i in s2:
    print(i)
```

```python
#Dictionary Data Type
d = {1: 'Geeks', 2: 'For', 3: 'Geeks'}
print(d[1])
print(d.get(2))
```
